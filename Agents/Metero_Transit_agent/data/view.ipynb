```python
import gtfs_kit as gk

# Load feed directly from the ZIP file
# Specify units: 'km', 'm', 'mi', or 'ft' for distance calculations
feed = gk.read_feed("dmrc_static_gtfs_v1.zip", dist_units="km")

# Quick health/validity check on the feed
# print("Is feed valid?:", feed.validate().empty)

# 1. Stops / Stations DataFrame
print("=== STOPS ===")
print(feed.stops[["stop_id", "stop_name", "stop_lat", "stop_lon"]].head())

# 2. Routes DataFrame (Yellow, Blue, Red Lines)
print("\n=== ROUTES ===")
print(feed.routes[["route_id", "route_short_name", "route_long_name"]].head())

# 3. Scheduled Stop Times
print("\n=== STOP TIMES ===")
print(
    feed.stop_times[["trip_id", "arrival_time", "stop_id", "stop_sequence"]].head()
)

# 4. Active Trips
print("\n=== TRIPS ===")
print(feed.trips[["trip_id", "route_id", "service_id"]].head())
```
